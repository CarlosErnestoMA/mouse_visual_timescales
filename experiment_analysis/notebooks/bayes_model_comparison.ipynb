```python
# precompute LOO for all models and stimuli

from itertools import product
from tqdm.notebook import tqdm

models = ['sgm', 'lmm']
measures = ["tau_double", "tau_R", "R_tot"]
stimuli = ["natural_movie_one_more_repeats", "natural_movie_three", "spontaneous",]

combinations = list(product(measures, stimuli, models))
loos = dict()

for meas, stim, model in tqdm(combinations):
    idata = az.from_netcdf(f"{data_dir}/bayes_cortex_{model}_{meas}_{stim}.nc")
    loos[f"{model}_{meas}_{stim}"] = az.loo(idata, pointwise=True)
```

```python
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
from bitsandbobs.plt import set_size

matplotlib.rcParams["savefig.facecolor"] = (0.0, 0.0, 0.0, 0.0)  # transparent figure bg
matplotlib.rcParams["axes.facecolor"] = (1.0, 0.0, 0.0, 0.0)
matplotlib.rcParams["xtick.labelsize"]= 6
matplotlib.rcParams["ytick.labelsize"]= 6


# stim = "spontaneous"
# stim = "natural_movie_three"
# stim = "natural_movie_one_more_repeats"

for stim in stimuli:
    _, ax = plt.subplots();
    for mdx, meas in enumerate(measures):
        # this is the same what arviz.compare does, but I decided to not use az.compare.
        # because it ranks the models, but we always want to have "linear_model - structure_groups".
        # 'loo_i' gives us the array of the leave-one-out samples, with length rows in df
        diff = (
            loos[f"lmm_{meas}_{stim}"]["loo_i"].values
            - loos[f"sgm_{meas}_{stim}"]["loo_i"].values
        )
        ax.errorbar(
            x=mdx + 1,
            y=np.sum(diff),
            yerr=np.sqrt(len(diff) * np.var(diff)),
            label=f"{meas}",
            fmt = "s",
            # markeredgewidth = 0.0,
            markersize = 3.0,
            markerfacecolor = 'white',
            ecolor = 'black',
            color = "black",
            elinewidth = 1.2,
            capsize = 2,
            clip_on=False,
        )
        print(f"{stim=} {meas=}, {np.sum(diff)=:.2f}")

    ax.set_xticks([])
    ax.set_ylim(-4.5, None)
    # yticks = [25, 20, 15, 10, 5, 0]
    # ax.set_yticks(yticks)
    if stim == "natural_movie_one_more_repeats":
        ax.set_ylim(-1, 21)
        yticks = [20, 15, 10, 5, 0]
        ax.set_yticks(yticks)
    for y in ax.get_yticks():
        ax.axhline(y, ls="-", lw=1.0 if y == 0 else 0.7, zorder=0, color="#E6E6E6")
    ax.set_xlim(0, 4)
    sns.despine(ax=ax, trim=False, bottom=True)
    # ax.set_ylabel("ELPD difference")
    ax.spines['left'].set_visible(False)
    ax.yaxis.set_ticks_position('none')

    set_size(ax, w=1.0, h=3.0, b=0.1, l=0.7, t=0.1, r=0.5)
    os.makedirs('../img', exist_ok=True)

    if stim == "natural_movie_one_more_repeats":
        text_kwargs = dict(
            fontsize=8,
            color="black",
            transform=ax.transAxes,
        )
        ax.text(0.7, 0.85, ha="right", va="bottom", s=r"$\tau_{\rm {c}}$", **text_kwargs)
        ax.text(1.03, 0.5, ha="right", va="bottom", s=r"$\tau_{\rm {R}}$", **text_kwargs)
        ax.text(1.45, 0.14, ha="right", va="bottom", s=r"$R_{\rm {tot}}$", **text_kwargs)

    if stim == "natural_movie_three":
        text_kwargs = dict(
            fontsize=8,
            color="black",
            transform=ax.transAxes,
        )
        ax.text(
            0.7, 0.88, ha="right", va="bottom", s=r"$\tau_{\rm {c}}$", **text_kwargs
        )
        ax.text(
            1.03, 0.53, ha="right", va="bottom", s=r"$\tau_{\rm {R}}$", **text_kwargs
        )
        ax.text(
            1.45, 0.07, ha="right", va="bottom", s=r"$R_{\rm {tot}}$", **text_kwargs
        )

    if stim == "spontaneous":
        text_kwargs = dict(
            fontsize=8,
            color="black",
            transform=ax.transAxes,
        )
        ax.text(
            0.7, 0.44, ha="right", va="bottom", s=r"$\tau_{\rm {c}}$", **text_kwargs
        )
        ax.text(
            1.03, 0.7, ha="right", va="bottom", s=r"$\tau_{\rm {R}}$", **text_kwargs
        )
        ax.text(
            1.45, 0.09, ha="right", va="bottom", s=r"$R_{\rm {tot}}$", **text_kwargs
        )

    ax.get_figure().savefig(f"../img/model_comparison_{stim}_all_measures.pdf")
```
